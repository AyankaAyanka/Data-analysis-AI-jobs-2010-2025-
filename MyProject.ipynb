```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\user\Desktop\ai_impact_jobs_2010_2025.csv")
print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
print(df.isnull().sum())
df = df.dropna()
print(df.duplicated().sum())
df = df.drop_duplicates()
df.dtypes
```
